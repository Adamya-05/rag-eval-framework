```python
print("="*60)
print("CONFIGURATION")
print("="*60)

CHUNK_SIZES = [300, 500]
CHUNK_OVERLAPS = [50]
EMBEDDING_MODELS = [
    "sentence-transformers/all-MiniLM-L6-v2",
    "sentence-transformers/all-mpnet-base-v2"
]
VECTOR_DBS = ["faiss", "chroma"]
TOP_K = 3

print("✓ Config loaded")
print(f"Chunk sizes      : {CHUNK_SIZES}")
print(f"Embedding models : {len(EMBEDDING_MODELS)}")
print(f"Vector DBs       : {VECTOR_DBS}")
```

```python
from langchain_community.document_loaders import PyPDFLoader
import os

print("\n" + "="*60)
print("DOCUMENT LOADING")
print("="*60)

docs = []

for file in os.listdir("/content"):
    if file.endswith(".pdf"):
        loader = PyPDFLoader(os.path.join("/content", file))
        loaded = loader.load()
        docs.extend(loaded)
        print(f"✓ Loaded {file} → {len(loaded)} pages")

print(f"\nTotal pages loaded: {len(docs)}")
```

```python
import re

def clean_text(text):
    text = re.sub(r'\n+', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

print("\n✓ Cleaning sample text...")

sample = docs[0].page_content
cleaned = clean_text(sample)

print("\n--- BEFORE ---")
print(sample[:300])

print("\n--- AFTER ---")
print(cleaned[:300])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

print("\n" + "="*60)
print("CHUNKING")
print("="*60)

chunk_results = {}

for size in CHUNK_SIZES:
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=size,
        chunk_overlap=50
    )

    chunks = splitter.split_documents(docs)
    chunk_results[size] = chunks

    print(f"✓ Chunk size {size} → {len(chunks)} chunks")

print("\nChunking complete ✓")
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

print("\n" + "="*60)
print("EMBEDDINGS")
print("="*60)

embedding_models = {}

for model in EMBEDDING_MODELS:
    print(f"Loading embedding model: {model}")
    embedding_models[model] = HuggingFaceEmbeddings(model_name=model)
    print(f"✓ Loaded {model}")
```

```python
from langchain_community.vectorstores import FAISS, Chroma

print("\n" + "="*60)
print("VECTOR DATABASE")
print("="*60)

vectorstores = {}

for model_name, emb in embedding_models.items():
    for size, chunks in chunk_results.items():

        model_short = model_name.split("/")[-1]
        key = f"{model_short}_{size}"

        print(f"\nCreating FAISS for {key}")
        vectorstores[f"faiss_{key}"] = FAISS.from_documents(chunks, emb)
        print("✓ FAISS created")

        print(f"Creating Chroma for {key}")

        # 🔥 FIX: unique collection name
        collection_name = f"chroma_{model_short}_{size}"

        vectorstores[f"chroma_{key}"] = Chroma.from_documents(
            chunks,
            emb,
            collection_name=collection_name
        )

        print("✓ Chroma created")
```

```python
print("\n" + "="*60)
print("RETRIEVAL TEST")
print("="*60)

query = "What is normalization in DBMS?"

for name, db in vectorstores.items():
    print(f"\n--- {name} ---")

    results = db.similarity_search(query, k=TOP_K)

    for i, r in enumerate(results):
        print(f"\nResult {i+1}:")
        print(r.page_content[:200])
```

```python
from groq import Groq

client = Groq(api_key="Insert Your key")

def generate_answer(model_name, context, query):

    print(f"\n🔹 Generating answer using: {model_name}")

    prompt = f"""
    You are a DBMS expert.

    Answer the question using the context below.
    If partial information is available, try to answer.

    Context:
    {context}

    Question:
    {query}
    """

    response = client.chat.completions.create(
        model=model_name,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3
    )

    answer = response.choices[0].message.content

    print("✓ Response generated")

    return answer
```

```python
print("\n" + "="*60)
print("FINAL RAG OUTPUT")
print("="*60)

query = "Explain normalization in DBMS"

db = list(vectorstores.values())[0]

results = db.similarity_search(query, k=5)

print("\n🔍 Retrieved Context:")
for i, r in enumerate(results):
    print(f"\nChunk {i+1}:\n{r.page_content[:200]}")

context = "\n".join([r.page_content for r in results])

models = ["llama-3.1-8b-instant", "openai/gpt-oss-120b"]

for model in models:
    answer = generate_answer(model, context, query)

    print("\n" + "-"*50)
    print(f"MODEL: {model}")
    print("-"*50)
    print(answer)
```

```python
def faithfulness_score(answer, context):
    answer_words = set(answer.lower().split())
    context_words = set(context.lower().split())

    overlap = len(answer_words & context_words)
    score = overlap / len(answer_words)

    return score

score = faithfulness_score(answer, context)
print("📊 Faithfulness:", round(score, 3))
```
